# Week 5 — Data Visualization and Communication

Visual storytelling using the Wine Recognition dataset. Each visualization answers a question and is followed by an interpretation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine
wine = load_wine(as_frame=True)
df = wine.frame.copy()
df["wine_class"] = df["target"].map({0:"class_0", 1:"class_1", 2:"class_2"})
df = df.drop(columns=["target"])

display(df.head())
print("Rows/columns:",df.shape)
```

## Story 1 — How many samples are in each class?

```python
counts=df["wine_class"].value_counts().sort_index()
counts.plot(kind="bar",figsize=(7,4),title="Wine Samples by Class")
plt.xlabel("Wine class"); plt.ylabel("Sample count"); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
display(counts)
```

Interpretation: Class counts show how the observations are distributed. The classes are not identical in size, so stratified splitting and macro-averaged evaluation metrics are appropriate.

## Story 2 — How does alcohol vary by class?

```python
plt.figure(figsize=(7,4))
sns.histplot(data=df,x="alcohol",hue="wine_class",bins=12,element="step",common_norm=False)
plt.title("Alcohol Distribution by Wine Class"); plt.xlabel("Alcohol"); plt.ylabel("Frequency"); plt.tight_layout(); plt.show()
```

Interpretation: Compare the centers and overlap of the distributions. Separation may make alcohol informative for classification, while overlap shows why one feature alone may not be sufficient.

## Story 3 — How are chemical measurements correlated?

```python
numeric=df.drop(columns="wine_class")
plt.figure(figsize=(11,8)); sns.heatmap(numeric.corr(),cmap="coolwarm",center=0,vmin=-1,vmax=1)
plt.title("Correlation Between Chemical Features"); plt.tight_layout(); plt.show()
```

Interpretation: Correlation highlights linear relationships between pairs of measurements. It does not establish causation and correlated predictors can complicate interpretation.

## Story 4 — Does color intensity vary by class?

```python
plt.figure(figsize=(7,4))
sns.boxplot(data=df,x="wine_class",y="color_intensity")
plt.title("Color Intensity by Wine Class"); plt.xlabel("Wine class"); plt.ylabel("Color intensity"); plt.tight_layout(); plt.show()
display(df.groupby("wine_class")["color_intensity"].agg(["mean","median","std","count"]))
```

Interpretation: Boxplots compare medians, spread and potential outliers across classes. They communicate distribution differences without assuming that the mean alone describes each group.

## Story 5 — Strongest pairwise correlations

```python
corr=numeric.corr()
pairs=corr.where(np.triu(np.ones(corr.shape),k=1).astype(bool)).stack()
display(pairs.reindex(pairs.abs().sort_values(ascending=False).index).head(10).to_frame("correlation"))
```
